QUESTÃO 3

```python
frase = input("Digite uma frase: ")

nova_frase = ""

for letra in frase:
  if letra.lower() not in "aeiou ":
    nova_frase += "9"
  else:
    nova_frase += letra

print(nova_frase)
```

QUESTÃO 4

```python
numeros = (input("Digite os números: ").split())
numeros = [int(numero) for numero in numeros]
nova_sequencia = []

for numero in numeros:
  if numero > 10 and numero % 2 != 0:
    nova_sequencia.append(numero)

print(nova_sequencia)

```
